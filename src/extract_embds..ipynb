```python
from embedding_api import EmbeddingModel, Preprocessor

data = [
    '{"id": 1, "body": "Hello", "tree": []}',
    '{"id": 2, "body": "Another post", "tree": []}',
    '{"id": 3, "body": "Final", "tree": []}',
]


preprocess = Preprocessor()
model = EmbeddingModel(
    ckpt_path="src/logs/pretrain/runs/2025-10-23_22-41-28/checkpoints/last.ckpt",
    device="cuda",
)


for item in data:
    graph = preprocess(item)
    emb = model.encode(graph)
    print("Single embedding:", emb.shape)


batch_embs = model.batch_encode(data, preprocess)
print("Batch output:", batch_embs.shape)
```
